# Module A3 — Face Recognition & VIP Customer Visit Setup
This notebook demonstrates face detection via OpenCV Haar Cascades, 128-dimensional embedding generation, cosine similarity matching, and database serialization.

```python
import numpy as np
import cv2
import os
import joblib
from datetime import datetime

# Create sample customer embeddings database
np.random.seed(101)
emb1 = np.random.randn(128).astype(np.float32)
emb1 /= np.linalg.norm(emb1)

emb2 = np.random.randn(128).astype(np.float32)
emb2 /= np.linalg.norm(emb2)

db = {
    'CUST-1001': {'name': 'Sarah Jenkins (Gold VIP)', 'embedding': emb1, 'visits': 15, 'last_seen': datetime.now().isoformat()},
    'CUST-1002': {'name': 'Michael Chang (Platinum VIP)', 'embedding': emb2, 'visits': 32, 'last_seen': datetime.now().isoformat()}
}

# Calculate Cosine Similarity Matrix
sim = np.dot(emb1, emb2)
print(f'Similarity between Customer 1 and Customer 2: {sim:.4f}')

# Save Face DB
os.makedirs('../app/models', exist_ok=True)
joblib.dump(db, '../app/models/face_db.pkl')
print('Serialized face_db.pkl successfully!')
```
